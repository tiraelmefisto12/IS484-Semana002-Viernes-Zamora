```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22.0, margen=1.0):
    diferencia = temperatura_actual - temperatura_objetivo
    if diferencia > margen:
        return "Encender Aire Acondicionado"
    elif diferencia < -margen:
        return "Encender Calefaccion"
    else:
        return "Apagar sistemas / Standby"

# Prueba del agente con la secuencia del laboratorio
percepciones = [18.0, 19.5, 21.8, 24.3, 26.0, 23.0, 20.5]
for t in percepciones:
    print(f"Percepción: {t}°C -> Acción: {agente_termostato(t)}")
```

## Sección 5.6: Reflexión sobre la Racionalidad del Agente

**1. ¿El agente termostato es racional?**
Sí, es un agente racional porque, según la percepción que recibe (temperatura actual), selecciona la acción adecuada para cumplir su objetivo de mantener el ambiente cerca de los 22°C (dentro del margen de ±1°C).

**2. ¿Bajo qué condiciones dejaría de ser racional?**
- **Falla del sensor:** Si el sensor mide datos incorrectos (por ejemplo, marca 15°C cuando en realidad hace 30°C), el agente tomará decisiones erróneas[cite: 1].
- **Margen demasiado amplio:** Si el margen fuera de ±10°C, el agente no activaría la climatización aunque el ambiente esté extremadamente frío o caliente, dejando de cumplir su medida de desempeño[cite: 1].
